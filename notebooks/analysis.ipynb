# 🎓 Student Academic Performance Prediction
### End-to-End Machine Learning Regression Project

**Objective**: Predict a student's expected final score (0–100) using historical academic records, study habits, and behavioral attributes.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Set chart styling
sns.set_theme(style="whitegrid")
%matplotlib inline
```

## 1. Load & Inspect the Dataset

```python
df = pd.read_csv('../dataset/student_performance.csv')
print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
print("\nSummary Statistics:")
df.describe().round(2)
```

## 2. Exploratory Data Analysis (EDA)

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['final_score'], kde=True, color='indigo', bins=25)
plt.title('Final Score Distribution')
plt.xlabel('Final Score (/100)')
plt.ylabel('Student Count')
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap='Blues', fmt='.2f', square=True)
plt.title('Correlation Heatmap')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.regplot(data=df, x='study_hours', y='final_score', ax=axes[0], color='teal')
axes[0].set_title('Study Hours vs Final Score')

sns.regplot(data=df, x='attendance', y='final_score', ax=axes[1], color='coral')
axes[1].set_title('Attendance vs Final Score')
plt.tight_layout()
plt.show()
```

## 3. Feature Selection & Train-Test Split

```python
features = ['study_hours', 'attendance', 'previous_score', 'assignment_score',
            'internal_marks', 'sleep_hours', 'participation', 'parental_support', 'extracurricular']
target = 'final_score'

X = df[features]
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
print(f"Training instances: {X_train.shape[0]}, Testing instances: {X_test.shape[0]}")
```

## 4. Train Multiple ML Regression Models

```python
models = {
    'Linear Regression': LinearRegression(),
    'Decision Tree': DecisionTreeRegressor(max_depth=6, random_state=42),
    'Random Forest': RandomForestRegressor(n_estimators=150, max_depth=10, random_state=42)
}

results = []
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    results.append({
        'Model': name,
        'MAE': round(mae, 3),
        'MSE': round(mse, 3),
        'RMSE': round(rmse, 3),
        'R2 Score': round(r2, 4)
    })

results_df = pd.DataFrame(results)
results_df
```

## 5. Feature Importance Analysis

```python
rf = models['Random Forest']
feat_imp = pd.Series(rf.feature_importances_, index=features).sort_values(ascending=False)

plt.figure(figsize=(8, 4))
feat_imp.plot(kind='barh', color='navy')
plt.title('Random Forest Feature Importance')
plt.xlabel('Importance')
plt.gca().invert_yaxis()
plt.show()
```
